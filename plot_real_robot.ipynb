# Real-robot validation — Mixed scenario, 20 trials

Bar plots of SR / NT / PL from the physical-robot trials.
Numbers come directly from `tab:real_robot_results` in the paper.

```python
import matplotlib.pyplot as plt
import numpy as np

# (method, SR%, CR%, NT_mean, NT_sd, PL_mean, PL_sd, ITR_mean, ITR_sd, SD_mean, SD_sd)
# SR and CR have no per-trial SD in the table — just the aggregate percentages.
ROWS = [
    ('GenSafeNav\n(Cons.)',  100.0, 0.0,  12.84, 1.63,  9.45, 1.90, 5.56, 3.67, 1.00, 0.15),
    ('GenSafeNav\n(Naive)',   80.0, 20.0, 10.54, 0.63,  6.52, 0.61, 4.82, 5.92, 1.00, 0.26),
    ('Ours',                 100.0, 0.0,  10.86, 0.60,  6.95, 0.31, 4.30, 4.42, 1.05, 0.15),
]
methods = [r[0] for r in ROWS]
sr      = np.array([r[1] for r in ROWS])
nt_m    = np.array([r[3] for r in ROWS])
nt_s    = np.array([r[4] for r in ROWS])
pl_m    = np.array([r[5] for r in ROWS])
pl_s    = np.array([r[6] for r in ROWS])

# Color: highlight Ours.
COLORS = ['#4C72B0', '#DD8452', '#55A868']  # cons, naive, ours

for name, *vals in ROWS:
    print(name.replace('\n', ' '), vals)
```

```python
# Reset any rcParams left over from prior runs of this cell.
plt.rcdefaults()

fig, axes = plt.subplots(1, 3, figsize=(13, 4.2))
x = np.arange(len(methods))
width = 0.6

# --- SR (no SD on table — flat bars) ---
ax = axes[0]
bars = ax.bar(x, sr, width, color=COLORS, edgecolor='black', linewidth=0.6)
ax.set_ylabel('Success Rate (%)')
ax.set_title('SR  (↑)')
ax.set_ylim(0, 110)
ax.set_xticks(x); ax.set_xticklabels(methods)
ax.grid(axis='y', alpha=0.3)
for b, v in zip(bars, sr):
    ax.text(b.get_x() + b.get_width()/2, v + 1.5, f'{v:.0f}%',
            ha='center', va='bottom', fontsize=12)

# --- NT ---
ax = axes[1]
bars = ax.bar(x, nt_m, width, yerr=nt_s, capsize=5,
              color=COLORS, edgecolor='black', linewidth=0.6,
              error_kw=dict(ecolor='black', lw=1.2))
ax.set_ylabel('Navigation Time (s)')
ax.set_title('NT  (↓)')
ax.set_ylim(0, max(nt_m + nt_s) * 1.2)
ax.set_xticks(x); ax.set_xticklabels(methods)
ax.grid(axis='y', alpha=0.3)
for b, m, s in zip(bars, nt_m, nt_s):
    ax.text(b.get_x() + b.get_width()/2, m + s + 0.3, f'{m:.2f}±{s:.2f}',
            ha='center', va='bottom', fontsize=11)

# --- PL ---
ax = axes[2]
bars = ax.bar(x, pl_m, width, yerr=pl_s, capsize=5,
              color=COLORS, edgecolor='black', linewidth=0.6,
              error_kw=dict(ecolor='black', lw=1.2))
ax.set_ylabel('Path Length (m)')
ax.set_title('PL  (↓)')
ax.set_ylim(0, max(pl_m + pl_s) * 1.2)
ax.set_xticks(x); ax.set_xticklabels(methods)
ax.grid(axis='y', alpha=0.3)
for b, m, s in zip(bars, pl_m, pl_s):
    ax.text(b.get_x() + b.get_width()/2, m + s + 0.2, f'{m:.2f}±{s:.2f}',
            ha='center', va='bottom', fontsize=11)

fig.suptitle('Real-robot validation — Mixed scenario, 20 trials', fontsize=13)
fig.tight_layout()
plt.show()
```

```python
# Interaction-shift benchmark (tab:ood_results) — 5 separate 1×3 figures
# (Non-aware / Aware / Mixed / Spatial Clusters / Average), one per scenario,
# each laid out exactly like the real-robot strip above.
#
# Only the same three methods as the real-robot table: GenSafeNav (Cons.),
# GenSafeNav (Naive), Ours.

# Per scenario: list of (method, CR%, CR_sd, NT_m, NT_sd, PL_m, PL_sd).
# Numbers verbatim from tab:ood_results in the paper.
OOD = {
    'Non-aware':        [('GenSafeNav\n(Cons.)',  2.80, 0.40, 16.59, 0.47, 23.53, 0.62),
                         ('GenSafeNav\n(Naive)', 40.50, 2.47, 11.61, 0.20, 16.88, 0.29),
                         ('Ours',                 2.96, 0.36, 16.51, 0.44, 23.44, 0.59)],
    'Aware':            [('GenSafeNav\n(Cons.)',  0.00, 0.00, 16.24, 0.47, 23.49, 0.64),
                         ('GenSafeNav\n(Naive)',  0.74, 0.38, 10.69, 0.07, 17.88, 0.16),
                         ('Ours',                 0.72, 0.33, 10.94, 0.11, 18.16, 0.28)],
    'Mixed':            [('GenSafeNav\n(Cons.)',  1.36, 0.73, 16.61, 0.58, 23.79, 0.74),
                         ('GenSafeNav\n(Naive)', 17.60, 4.79, 11.36, 0.20, 17.82, 0.41),
                         ('Ours',                 4.48, 1.53, 13.63, 0.40, 20.64, 0.53)],
    'Spatial Clusters': [('GenSafeNav\n(Cons.)',  2.88, 0.77, 15.47, 0.68, 22.40, 0.89),
                         ('GenSafeNav\n(Naive)', 17.10, 2.43, 10.83, 0.10, 17.43, 0.33),
                         ('Ours',                 4.64, 0.92, 12.85, 0.30, 19.83, 0.58)],
}

# Build an "Average" entry: per-method, mean of the four scenario means/sds.
scenario_order = ['Non-aware', 'Aware', 'Mixed', 'Spatial Clusters']
_methods       = [r[0] for r in OOD['Non-aware']]
avg_rows = []
for i, m in enumerate(_methods):
    cols = np.array([OOD[s][i][1:] for s in scenario_order])  # (4, 6)
    avg_rows.append((m, *cols.mean(axis=0).tolist()))
OOD['Average'] = avg_rows

plot_order = scenario_order + ['Average']
CR_YMAX = 25.0  # CR panel capped — Non-aware Naive (40.5%) will be clipped.

for scenario in plot_order:
    rows = OOD[scenario]
    m_names = [r[0] for r in rows]
    cr_m  = np.array([r[1] for r in rows]); cr_s  = np.array([r[2] for r in rows])
    nt_m  = np.array([r[3] for r in rows]); nt_s  = np.array([r[4] for r in rows])
    pl_m  = np.array([r[5] for r in rows]); pl_s  = np.array([r[6] for r in rows])

    fig, axes = plt.subplots(1, 3, figsize=(13, 4.2))
    x = np.arange(len(m_names))
    width = 0.6

    # --- CR ---
    ax = axes[0]
    bars = ax.bar(x, cr_m, width, yerr=cr_s, capsize=5,
                  color=COLORS, edgecolor='black', linewidth=0.6,
                  error_kw=dict(ecolor='black', lw=1.2))
    ax.set_ylabel('Collision Rate (%)')
    ax.set_title('CR  (↓)')
    ax.set_ylim(0, CR_YMAX)
    ax.set_xticks(x); ax.set_xticklabels(m_names)
    ax.grid(axis='y', alpha=0.3)
    for b, mv, sv in zip(bars, cr_m, cr_s):
        # Keep annotation visible even when the bar exceeds CR_YMAX.
        y = min(mv + sv + 0.6, CR_YMAX - 1.8)
        ax.text(b.get_x() + b.get_width()/2, y, f'{mv:.2f}%',
                ha='center', va='bottom', fontsize=12)

    # --- NT ---
    ax = axes[1]
    bars = ax.bar(x, nt_m, width, yerr=nt_s, capsize=5,
                  color=COLORS, edgecolor='black', linewidth=0.6,
                  error_kw=dict(ecolor='black', lw=1.2))
    ax.set_ylabel('Navigation Time (s)')
    ax.set_title('NT  (↓)')
    ax.set_ylim(0, max(nt_m + nt_s) * 1.2)
    ax.set_xticks(x); ax.set_xticklabels(m_names)
    ax.grid(axis='y', alpha=0.3)
    for b, mv, sv in zip(bars, nt_m, nt_s):
        ax.text(b.get_x() + b.get_width()/2, mv + sv + 0.3, f'{mv:.2f}±{sv:.2f}',
                ha='center', va='bottom', fontsize=11)

    # --- PL ---
    ax = axes[2]
    bars = ax.bar(x, pl_m, width, yerr=pl_s, capsize=5,
                  color=COLORS, edgecolor='black', linewidth=0.6,
                  error_kw=dict(ecolor='black', lw=1.2))
    ax.set_ylabel('Path Length (m)')
    ax.set_title('PL  (↓)')
    ax.set_ylim(0, max(pl_m + pl_s) * 1.2)
    ax.set_xticks(x); ax.set_xticklabels(m_names)
    ax.grid(axis='y', alpha=0.3)
    for b, mv, sv in zip(bars, pl_m, pl_s):
        ax.text(b.get_x() + b.get_width()/2, mv + sv + 0.2, f'{mv:.2f}±{sv:.2f}',
                ha='center', va='bottom', fontsize=11)

    fig.suptitle(f'Interaction-shift benchmark — {scenario}', fontsize=13)
    fig.tight_layout()
    plt.show()
```
